## Dataframe

```python
import pandas as pd
from langchain_google_genai import ChatGoogleGenerativeAI

df = pd.read_excel(
    "Logbook chinese.xlsx"
)
```

```python
agent = create_pandas_dataframe_agent(
    ChatGoogleGenerativeAI(temperature=0, model="gpt-3.5-turbo-0613"),
    df,
    verbose=True,
    agent_type=AgentType.OPENAI_FUNCTIONS,
    allow_dangerous_code=True
)
```

```python
# create a new Chat with OpenAI
llm = ChatGoogleGenerativeAI(temperature=0.7, model="gemini-2.5-flash")

# set up the conversation memory for the chat
memory = ConversationBufferMemory(memory_key='chat_history', return_messages=True)

# putting it together: set up the conversation chain with the GPT 4o-mini LLM, the vector store and memory
conversation_chain = ConversationalRetrievalChain.from_llm(llm=llm, memory=memory)
```
